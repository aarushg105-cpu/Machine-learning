```python
import pandas as pd
import numpy as np

df = pd.read_csv("../01-Intro/car_fuel_efficiency_2026.csv")

df = df[
    [
        "engine_displacement",
        "horsepower",
        "vehicle_weight",
        "model_year",
        "fuel_efficiency_mpg"
    ]
]

df.head()
```

```python
df.isna().sum()
```

```python
median = df["horsepower"].median()
```

```python
n= len(df)
n_val= int(n*0.2)
n_test= int (n*0.2)
n_train= n- n_val-n_test
np.random.seed(42)
indx=np.arange(n)
np.random.shuffle(indx)

df_train=df.iloc[indx[:n_train]]
df_val=df.iloc[indx[n_train:n_train+ n_val]]
df_test=df.iloc[indx[n_train+n_val:]]
```

```python
features = [
    "engine_displacement",
    "horsepower",
    "vehicle_weight",
    "model_year"
]

target = "fuel_efficiency_mpg"
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error

X_train = df_train[features].fillna(0)
y_train = df_train[target]

X_val = df_val[features].fillna(0)
y_val = df_val[target]

model = LinearRegression()
model.fit(X_train, y_train)

pred = model.predict(X_val)

rmse_zero = root_mean_squared_error(y_val, pred)
rmse_zero
```

```python
X_train = df_train[features].copy()
X_val = df_val[features].copy()

for col in features:
    mean = X_train[col].mean()
    X_train[col] = X_train[col].fillna(mean)
    X_val[col] = X_val[col].fillna(mean)

model = LinearRegression()
model.fit(X_train, y_train)

pred = model.predict(X_val)

rmse_mean = root_mean_squared_error(y_val, pred)
rmse_mean
rmse_zero, rmse_mean

```

```python
from sklearn.linear_model import Ridge

X_train = df_train[features].fillna(0)
X_val = df_val[features].fillna(0)

for r in [0, 0.01, 0.1, 1, 5, 10, 100]:
    model = Ridge(alpha=r)
    model.fit(X_train, y_train)
    pred = model.predict(X_val)
    rmse = root_mean_squared_error(y_val, pred)
    print("r=",r , "\t","RMSE=" ,round(rmse, 4))
```

```python
rmse_scores = []

for seed in range(10):
    n = len(df)
    n_val = int(n * 0.2)
    n_test = int(n * 0.2)
    n_train = n - n_val - n_test

    np.random.seed(seed)
    idx = np.arange(n)
    np.random.shuffle(idx)

    df_train = df.iloc[idx[:n_train]]
    df_val = df.iloc[idx[n_train:n_train + n_val]]

    X_train = df_train[features].fillna(0)
    y_train = df_train[target]

    X_val = df_val[features].fillna(0)
    y_val = df_val[target]

    model = LinearRegression()
    model.fit(X_train, y_train)

    pred = model.predict(X_val)

    rmse = root_mean_squared_error(y_val, pred)
    rmse_scores.append(rmse)

rmse_scores
```

```python
round(np.std(rmse_scores), 3)
```
